# Vector Operations in Python with NumPy

This notebook demonstrates fundamental vector operations used in machine learning mathematics. We'll explore vector creation, arithmetic operations, dot products, norms, and visualization using NumPy and Matplotlib.

```python
# Import required libraries
import numpy as np
import matplotlib.pyplot as plt
```

## Creating a Vector

We create our first vector using NumPy's `array` function. A vector is a 1-dimensional array representing magnitude and direction in space.

```python
# Create a 2D vector with components [3, 4]
vector_u = np.array([3, 4])
vector_u
```

### Displaying Vector Components

Using `print()` to display the vector's components in a clean format.

```python
print(vector_u)
```

### Checking Vector Data Type

The `type()` function confirms our vector is a NumPy ndarray (n-dimensional array).

```python
type(vector_u)
```

### Vector Dimension (Length)

The `len()` function returns the number of components (dimension) of the vector.

```python
len(vector_u)
```

### Accessing Individual Components

Vector components are accessed using zero-based indexing: `vector[0]` for the first component, `vector[1]` for the second.

```python
# Access individual components
vector_u[0], vector_u[1]
```

## Creating a Second Vector

We create another 2D vector for demonstrating vector operations.

```python
# Create second vector with components [7, 2]
vector_v = np.array([7, 2])
print(vector_v)
```

### Displaying Both Vectors

Printing both vectors together for comparison.

```python
print(vector_u, vector_v)
```

## Vector Addition

Element-wise addition: `vector_u + vector_v` adds corresponding components.
Mathematically: \( \vec{u} + \vec{v} = [u_1 + v_1, u_2 + v_2] \)

```python
# Vector addition
vector_sum = vector_u + vector_v
vector_sum
```

## Vector Subtraction

Element-wise subtraction: `vector_u - vector_v` subtracts corresponding components.
Mathematically: \( \vec{u} - \vec{v} = [u_1 - v_1, u_2 - v_2] \)

```python
# Vector subtraction
vector_diff = vector_u - vector_v
vector_diff
```

## Element-wise Multiplication (Hadamard Product)

Element-wise multiplication: `vector_u * vector_v` multiplies corresponding components.
This is also called the Hadamard product.
Mathematically: \( \vec{u} \odot \vec{v} = [u_1 \cdot v_1, u_2 \cdot v_2] \)

```python
# Element-wise multiplication (Hadamard product)
hadamard_product = vector_u * vector_v
hadamard_product
```

### Using np.multiply()

NumPy's `multiply()` function performs the same element-wise multiplication as the `*` operator.

```python
# Alternative way: np.multiply()
np.multiply(vector_u, vector_v)
```

## Scalar Multiplication

Multiplying a vector by a scalar scales each component.
Mathematically: \( c \cdot \vec{u} = [c \cdot u_1, c \cdot u_2] \)

```python
# Scalar multiplication: scale vector_u by 2
scaled_vector = np.dot(2, vector_u)
scaled_vector
```

### Scalar Dot Product (Scalar × Scalar)

When `np.dot()` is used with two scalars, it returns their simple product.

```python
# Dot product of two scalars
np.dot(5, 4)
```

## Dot Product (Inner Product)

The dot product of two vectors is a scalar computed as:

\( \vec{u} \cdot \vec{v} = u_1 v_1 + u_2 v_2 = 3 \cdot 7 + 4 \cdot 2 = 21 + 8 = 29 \)

Geometrically: \( \vec{u} \cdot \vec{v} = ||\vec{u}|| \cdot ||\vec{v}|| \cdot \cos(\theta) \)

```python
# Dot product of two vectors
dot_product = np.dot(vector_u, vector_v)
dot_product
```

## Vector Norm (Magnitude/Length)

The Euclidean norm (L2 norm) represents the vector's length:

\( ||\vec{u}|| = \sqrt{u_1^2 + u_2^2} = \sqrt{3^2 + 4^2} = \sqrt{9 + 16} = \sqrt{25} = 5 \)

```python
# Euclidean norm (magnitude) of vector_u
norm_u = np.linalg.norm(vector_u)
norm_u
```

### Norm of a 3D Vector

The same `np.linalg.norm()` function works for vectors of any dimension.

```python
# Create a 3D vector and compute its norm
vector_3d = np.array([1, 2, 3])
norm_3d = np.linalg.norm(vector_3d)
norm_3d
```

## Visualizing a Single Vector

Using Matplotlib's `quiver` function to plot a 2D vector as an arrow from the origin.

```python
# Plot vector_u as an arrow
plt.quiver(0, 0, vector_u[0], vector_u[1], 
           angles='xy', scale_units='xy', scale=1, color='blue')
plt.xlim(0, 5)
plt.ylim(0, 5)
plt.axhline(y=0, color='grey', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='grey', linestyle='-', alpha=0.3)
plt.grid(True, alpha=0.3)
plt.gca().set_aspect('equal', adjustable='box')
plt.text(vector_u[0], vector_u[1], r'$\vec{u}$', 
         fontsize=20, ha='right', va='bottom', color='blue')
plt.title('Vector Visualization')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

## Visualizing Vector Addition

The sum of two vectors is visualized by placing them head-to-tail. The resultant vector (red) goes from the origin to the tip of the second vector.

```python
# Vector addition visualization
vector_sum = vector_u + vector_v

# Plot vector_u
plt.quiver(0, 0, vector_u[0], vector_u[1], 
           angles='xy', scale_units='xy', scale=1, color='blue', label='$\vec{u}$')
# Plot vector_v starting from tip of vector_u (head-to-tail)
plt.quiver(vector_u[0], vector_u[1], vector_v[0], vector_v[1], 
           angles='xy', scale_units='xy', scale=1, color='green', label='$\vec{v}$')
# Plot resultant vector (sum)
plt.quiver(0, 0, vector_sum[0], vector_sum[1], 
           angles='xy', scale_units='xy', scale=1, color='red', label='$\vec{u} + \vec{v}$')

plt.xlim(0, 11)
plt.ylim(0, 7)
plt.axhline(y=0, color='grey', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='grey', linestyle='-', alpha=0.3)
plt.grid(True, alpha=0.3)
plt.gca().set_aspect('equal', adjustable='box')
plt.legend(fontsize=14)
plt.title('Vector Addition: Head-to-Tail Method')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```

## Visualizing Vector Subtraction

Vector subtraction \( \vec{u} - \vec{v} \) is equivalent to \( \vec{u} + (-\vec{v}) \). 
We visualize \( \vec{u} \), \( \vec{v} \), and their difference \( \vec{u} - \vec{v} \) (shown in red).

```python
# Vector subtraction visualization
vector_diff = vector_u - vector_v

# Plot vector_u
plt.quiver(0, 0, vector_u[0], vector_u[1], 
           angles='xy', scale_units='xy', scale=1, color='blue', label='$\vec{u}$')
# Plot vector_v
plt.quiver(0, 0, vector_v[0], vector_v[1], 
           angles='xy', scale_units='xy', scale=1, color='green', label='$\vec{v}$')
# Plot difference vector (u - v)
plt.quiver(0, 0, vector_diff[0], vector_diff[1], 
           angles='xy', scale_units='xy', scale=1, color='red', label='$\vec{u} - \vec{v}$')

plt.xlim(-5, 9)
plt.ylim(-1, 5)
plt.axhline(y=0, color='grey', linestyle='-', alpha=0.3)
plt.axvline(x=0, color='grey', linestyle='-', alpha=0.3)
plt.grid(True, alpha=0.3)
plt.gca().set_aspect('equal', adjustable='box')
plt.legend(fontsize=14)
plt.title('Vector Subtraction: $\vec{u} - \vec{v}$')
plt.xlabel('x')
plt.ylabel('y')
plt.show()
```
