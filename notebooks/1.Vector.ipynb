## 1. Mathematical Concept of Vectors

### Intuition First
In machine learning, data points (such as house prices with features like square footage and bedrooms, or word embeddings in NLP) are represented as vectors. Geometrically, a vector in $\mathbb{R}^n$ represents a point or a directed arrow in an $n$-dimensional vector space. Computable vector operations allow algorithms to measure similarity, compute distances, and apply linear models.

### Formal Definition / Math
A real column vector $v \in \mathbb{R}^n$ is an ordered tuple of $n$ real components:

$$\begin{equation}
v = \begin{bmatrix} v_1 \\ v_2 \\ \vdots \\ v_n \end{bmatrix} \in \mathbb{R}^n
\end{equation}$$

where:
- $n$ is the dimension (number of elements/features).
- $v_i \in \mathbb{R}$ represents the component at position $i$.

### Connection to the Code
In NumPy, 1D arrays `np.array([3, 4])` represent vectors of dimension $n=2$.

**Key Takeaway**: **Vectors structure multi-dimensional data points in feature space, serving as the foundational building block for ML algorithms.**

## 2. Vector Addition and Subtraction

### Intuition First
Adding two vectors combines their directional displacements component-by-component. Geometrically, vector addition follows the **triangle law** (or head-to-tail method). Vector subtraction calculates the displacement vector pointing from one vector tip to another.

### Formal Definition / Math
For vectors $u, v \in \mathbb{R}^n$, component-wise addition and subtraction are defined as:

$$\begin{equation}
u \pm v = \begin{bmatrix} u_1 \pm v_1 \\ u_2 \pm v_2 \\ \vdots \\ u_n \pm v_n \end{bmatrix}
\end{equation}$$

### Connection to the Code
- `vector_sum = vector_u + vector_v` computes $u + v$.
- `vector_diff = vector_u - vector_v` computes $u - v$.

### Worked Micro-Example
Given $u = [3, 4]^T$ and $v = [7, 2]^T$:
- $u + v = [3+7, 4+2]^T = [10, 6]^T$.
- $u - v = [3-7, 4-2]^T = [-4, 2]^T$.

**Key Takeaway**: **Vector addition and subtraction operate component-wise, translating geometric displacements in vector space.**

```python
vector_v = np.array([7, 2])
vector_sum = vector_u + vector_v
vector_diff = vector_u - vector_v

print(f"Vector v: {vector_v}")
print(f"Vector Sum (u + v): {vector_sum}")
print(f"Vector Difference (u - v): {vector_diff}")
```

## 3. Element-wise (Hadamard) & Scalar Multiplication

### Intuition First
Scalar multiplication stretches or shrinks a vector by scaling every component by constant $c \in \mathbb{R}$. Element-wise (Hadamard) multiplication multiplies corresponding elements of two vectors of identical shape, widely used in neural network gating mechanisms (e.g. LSTMs and attention masks).

### Formal Definition / Math
- **Scalar Multiplication**: $c \cdot u = [c \cdot u_1, c \cdot u_2, \dots, c \cdot u_n]^T$.
- **Hadamard Product**: $u \odot v = [u_1 v_1, u_2 v_2, \dots, u_n v_n]^T$.

### Connection to the Code
- `2 * vector_u` computes scalar multiplication $c \cdot u$.
- `vector_u * vector_v` or `np.multiply(vector_u, vector_v)` computes $u \odot v$.

**Key Takeaway**: **Scalar multiplication scales vector magnitude without altering its span line, while Hadamard multiplication scales individual features element-by-element.**

```python
scaled_u = 2 * vector_u
hadamard_prod = vector_u * vector_v

print(f"Scalar Multiplication (2 * u): {scaled_u}")
print(f"Hadamard Product (u * v): {hadamard_prod}")
```

## 4. Dot Product (Inner Product) & Vector Norms

### Intuition First
The dot product calculates a single scalar indicating how aligned two vectors are in space. If the dot product is zero, the vectors are orthogonal (perpendicular). Vector norms measure the 'length' or magnitude of a vector from the origin.

### Formal Definition / Math
- **Dot Product**: $u \cdot v = u^T v = \sum_{i=1}^n u_i v_i = \|u\|_2 \|v\|_2 \cos(\theta)$, where $\theta$ is the angle between $u$ and $v$.
- **Euclidean Norm ($\ell_2$ Norm)**: $\|u\|_2 = \sqrt{\sum_{i=1}^n u_i^2} = \sqrt{u \cdot u}$.

### Connection to the Code
- `np.dot(vector_u, vector_v)` calculates dot product $u \cdot v$.
- `np.linalg.norm(vector_u)` calculates Euclidean norm $\|u\|_2$.

### Worked Micro-Example
For $u = [3, 4]^T$:
- Dot product $u \cdot v = 3(7) + 4(2) = 21 + 8 = 29$.
- Norm $\|u\|_2 = \sqrt{3^2 + 4^2} = \sqrt{25} = 5.0$.

**Key Takeaway**: **The dot product measures directional alignment ($\|u\|\|v\|\cos\theta$), and the $\|u\|_2$ norm measures physical Euclidean length.**

```python
dot_prod = np.dot(vector_u, vector_v)
norm_u = np.linalg.norm(vector_u)
vector_3d = np.array([1, 2, 3])
norm_3d = np.linalg.norm(vector_3d)

print(f"Dot product (u . v): {dot_prod}")
print(f"Norm of u (||u||_2): {norm_u}")
print(f"3D vector norm: {norm_3d}")
```
