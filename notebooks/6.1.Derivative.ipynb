```python
import sympy as sp
```

```python
x = sp.Symbol("x")
y = sp.Symbol("y")
```

```python
f = x**3 + 2 * x + 3 + 2 * y
f
```

```python
sp.diff(f, x)
```

```python
sp.diff(f, y)
```

```python
f = sp.sin(x**2 + 5 * x)
f
```

```python
sp.diff(f)
```

```python
f = sp.log(x**2)
f
```

```python
sp.diff(f)
```

```python
f = sp.log(x**2 + 3 * x - 6)
f
```

```python
sp.diff(f)
```

```python
f = sp.exp(3 * x**2 - 4 * x + 9)
f
```

```python
sp.diff(f, x)
```
