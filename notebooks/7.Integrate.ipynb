## 1. Concept of Integration & Indefinite Integrals

### Intuition First
Integration is the inverse operation of differentiation (antidifferentiation). In machine learning, integration is used to calculate cumulative probabilities from continuous probability density functions (PDFs), compute expected values, and evaluate cumulative loss surfaces.

### Formal Definition / Math
The fundamental theorem of calculus states that if $F'(x) = f(x)$, then the indefinite integral of $f(x)$ is:

$$\begin{equation}
\int f(x) \, dx = F(x) + C
\end{equation}$$

where $C \in \mathbb{R}$ is the constant of integration.

#### Basic Power Rule for Integration:
For $n \neq -1$:

$$\begin{equation}
\int x^n \, dx = \frac{x^{n+1}}{n+1} + C
\end{equation}$$

### Connection to the Code
- `x = sp.Symbol('x')`: Defines continuous real symbol $x$.
- `sp.integrate(x, x)`: Computes $\int x \, dx = \frac{x^2}{2}$.
- `sp.integrate(x**2, x)`: Computes $\int x^2 \, dx = \frac{x^3}{3}$.

**Key Takeaway**: **Indefinite integration computes antiderivatives, satisfying $\int x^n \, dx = \frac{x^{n+1}}{n+1} + C$.**

```python
import sympy as sp

x = sp.Symbol("x")
y = sp.Symbol("y")
```

```python
sp.integrate(x)
```

```python
sp.integrate(x, x)
```

## 2. Partial Integration in Multivariate Spaces

### Intuition First
Just as partial differentiation holds non-target variables constant, partial integration integrates a multivariate function with respect to one variable while treating all other variables as constants. This forms the foundation for joint and marginal probability calculations in machine learning.

### Formal Definition / Math
For a bivariate function $f(x, y)$:
- Partial integration with respect to $x$:
  $$\int f(x, y) \, dx$$
- Partial integration with respect to $y$:
  $$\int f(x, y) \, dy$$

#### Derivation Steps for $f(x, y) = x^2 + 3y$:
1. Integrating with respect to $x$ (treating $y$ as a constant):
   $$\int (x^2 + 3y) \, dx = \frac{x^3}{3} + 3xy$$
2. Integrating with respect to $y$ (treating $x$ as a constant):
   $$\int (x^2 + 3y) \, dy = x^2 y + \frac{3y^2}{2}$$

### Connection to the Code
- `sp.integrate(x, y)`: Integrates $x$ with respect to $y$, yielding $xy$.
- `sp.integrate(f, x)`: Integrates $x^2 + 3y$ with respect to $x$.
- `sp.integrate(f, y)`: Integrates $x^2 + 3y$ with respect to $y$.

**Key Takeaway**: **Partial integration integrates multivariate expressions along a selected target axis while holding remaining variables constant.**

```python
sp.integrate(x, y)
```

```python
sp.integrate(x**2, y)
```

## 3. Transcendental & Exponential Integration

### Intuition First
Trigonometric and exponential integrals model periodic signals and decaying probability distributions (e.g., Laplace or exponential distributions in statistical learning).

### Formal Definition / Math
- Trigonometric integral: $\int \sin(x) \, dx = -\cos(x) + C$.
- Exponential integral: $\int e^{ax} \, dx = \frac{1}{a} e^{ax} + C$.

#### Derivation Steps for $e^{-2x}$:
$$\int e^{-2x} \, dx = \frac{1}{-2} e^{-2x} = -\frac{e^{-2x}}{2}$$

### Connection to the Code
- `sp.integrate(sp.sin(x), x)`: Yields `-cos(x)`.
- `sp.integrate(sp.exp(-2 * x), x)`: Yields `-exp(-2*x)/2`.
- `sp.integrate(sp.exp(2 * x), (x, 0, sp.oo))`: Improper integral evaluated from 0 to $\infty$, yielding divergence $\infty$.

**Key Takeaway**: **Exponential and trigonometric integrals evaluate decay rates and periodic functions in probabilistic ML.**

```python
sp.integrate(sp.sin(x), x)
```

```python
sp.integrate(sp.exp(-2 * x), x)
```

```python
sp.integrate(sp.exp(2 * x), (x, 0, sp.oo))
```

## 4. Definite Integration & Area Under Curves

### Intuition First
Definite integrals compute the net signed area under a curve between bounds $[a, b]$, corresponding to cumulative density or total accumulated quantity.

### Formal Definition / Math
By the Fundamental Theorem of Calculus:

$$\begin{equation}
\int_{a}^{b} f(x) \, dx = F(b) - F(a)
\end{equation}$$

#### Worked Micro-Example:
For $f(x) = x$ over interval $[1, 2]$:

$$\int_{1}^{2} x \, dx = \left[ \frac{x^2}{2} \right]_{1}^{2} = \frac{2^2}{2} - \frac{1^2}{2} = 2 - 0.5 = 1.5$$
