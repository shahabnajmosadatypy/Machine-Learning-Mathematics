```python
import sympy as sp
```

```python
x = sp.Symbol("x")
y = sp.Symbol("y")
```

```python
sp.integrate(x)
```

```python
sp.integrate(x, x)
```

```python
sp.integrate(x, y)
```

```python
sp.integrate(x**2, x)
```

```python
sp.integrate(x**2, y)
```

```python
f = x**2 + 3 * y
f
```

```python
sp.integrate(f, x)
```

```python
sp.integrate(f, y)
```

```python
sp.integrate(sp.sin(x), x)
```

```python
sp.integrate(sp.exp(-2 * x), x)
```

```python
sp.integrate(sp.exp(2 * x), (x, 0, sp.oo))
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(-3, 3, 1000)
plt.plot(x, x, color="r")
plt.axhline(color="gray")
plt.fill_between(x, x, where=[(i >= 1) and (i <= 2) for i in x], color="g")
plt.show()
```
