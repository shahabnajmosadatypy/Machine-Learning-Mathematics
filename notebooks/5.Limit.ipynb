```python
import sympy as sp
```

```python
x = sp.Symbol("x")
```

```python
# Just for showing how sympy works
x**2 + 5 + 4 * x + sp.log(x) + sp.exp(x) + (1 / x)
```

```python
f = (x**2 - 1) / (x - 1)
f
```

```python
sp.limit(f, x, 1)
```

```python
def f(x):
    y = (x**2 - 1) / (x - 1)
    return y
```

```python
f(0.9999)
```

```python
f(1.5)
```

```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x = np.linspace(0, 2, 10)
y = (x**2 - 1) / (x - 1)
plt.plot(x, y)
plt.axvline(x=1, color="r", linestyle="--")
plt.axhline(y=2, color="g", linestyle="--")
plt.show()
```

```python
x = sp.Symbol("x")
f = 1 / x
f
```

```python
sp.limit(f, x, sp.oo)
```
