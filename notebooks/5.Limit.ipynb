## 2. Formal Definition of Limits & Indeterminate Forms

### Intuition First
A limit answers the question: *As input $x$ gets arbitrarily close to a target point $c$, what value does $f(x)$ approach?* This concept allows us to evaluate functions at points where direct substitution yields undefined indeterminate forms such as $\frac{0}{0}$.

### Formal Definition / Math
The formal $(\epsilon, \delta)$-definition of a limit states that:

$$\begin{equation}
\lim_{x \to c} f(x) = L
\end{equation}$$

if for every $\epsilon > 0$, there exists a $\delta > 0$ such that $0 < |x - c| < \delta \implies |f(x) - L| < \epsilon$.

#### Derivation for Indeterminate Form $\frac{0}{0}$:
For $f(x) = \frac{x^2 - 1}{x - 1}$, direct evaluation at $x = 1$ gives $\frac{1^2 - 1}{1 - 1} = \frac{0}{0}$. Factoring the numerator yields:

$$\begin{equation}
f(x) = \frac{(x - 1)(x + 1)}{x - 1} = x + 1 \quad (x \neq 1)
\end{equation}$$

Taking the limit as $x \to 1$:

$$\begin{equation}
\lim_{x \to 1} \frac{x^2 - 1}{x - 1} = \lim_{x \to 1} (x + 1) = 2
\end{equation}$$

### Connection to the Code
- `f = (x**2 - 1) / (x - 1)` defines function $f(x)$.
- `sp.limit(f, x, 1)` analytically computes $\lim_{x \to 1} f(x) = 2$.

### Worked Micro-Example
At $x = 1.0001$: $f(1.0001) = 1.0001 + 1 = 2.0001 \approx 2$.

### Common Pitfalls / Notes
A function does **not** need to be defined at point $x = c$ for its limit $\lim_{x \to c} f(x)$ to exist. The limit describes the local behavior *near* $c$, not *at* $c$.

**Key Takeaway**: **Limits resolve indeterminate forms $\frac{0}{0}$ by analyzing function behavior infinitely close to a point.**

```python
f = (x**2 - 1) / (x - 1)
f
```

```python
# Symbolic computation of limit as x approaches 1
sp.limit(f, x, 1)
```

## 3. Numerical Approximation of Limits

### Intuition First
In numerical optimization algorithms, analytical limits are approximated by evaluating function values at very small increments $\Delta x$.

### Formal Definition / Math
As $x \to c^-$ (from the left) or $x \to c^+$ (from the right), the function values $f(c - \delta)$ and $f(c + \delta)$ converge to $L$.

### Connection to the Code
- `def f(x): return (x**2 - 1) / (x - 1)` defines numerical function.
- `f(0.9999)` evaluates left-hand approximation $f(1 - 10^{-4}) \approx 1.9999$.
- `f(1.5)` evaluates function at $x = 1.5$.

**Key Takeaway**: **Evaluating $f(x)$ at points near $x = c$ numerically demonstrates convergence toward limit value $L$.**

```python
def f(x):
    y = (x**2 - 1) / (x - 1)
    return y
```

```python
f(0.9999)
```

```python
f(1.5)
```

## 5. Limits at Infinity & Asymptotic Behavior

### Intuition First
In deep learning, limits at infinity describe asymptotic behavior—such as activation functions (e.g. sigmoid or softmax) flattening out or loss surfaces stabilizing as parameters $x \to \infty$.

### Formal Definition / Math
For reciprocal function $f(x) = \frac{1}{x}$, as $x$ increases without bound ($x \to \infty$), the output approaches 0:

$$\begin{equation}
\lim_{x \to \infty} \frac{1}{x} = 0
\end{equation}$$

The line $y = 0$ is a **horizontal asymptote** of $f(x)$.

### Connection to the Code
- `f = 1 / x`: Defines reciprocal function.
- `sp.limit(f, x, sp.oo)`: Computes limit as $x \to \infty$ (using SymPy infinity symbol `sp.oo`).
- Plot shows function decaying toward horizontal axis line $y = 0$.

**Key Takeaway**: **Limits at infinity describe horizontal asymptotes and steady-state long-term behavior.**

```python
x_sym = sp.Symbol("x")
f_reciprocal = 1 / x_sym
f_reciprocal
```

```python
# Compute limit as x approaches infinity
sp.limit(f_reciprocal, x_sym, sp.oo)
```

```python
# Plot reciprocal function limit at infinity
def recip(x):
    return 1.0 / x

fig_inf = plot_function_limit(
    func=recip,
    x_range=(1.0, 100.0),
    target_x=None,
    limit_val=0.0,
    title="Asymptotic Limit at Infinity: lim_{x -> oo} 1/x = 0"
)
plt.show()
```
