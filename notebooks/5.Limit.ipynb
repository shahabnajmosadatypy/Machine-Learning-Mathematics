```python
import sympy as sp
```

```python
x = sp.Symbol("x")
```

```python
# Just for showing how sympy works
x**2 + 5 + 4 * x + sp.log(x) + sp.exp(x) + (1 / x)
```

```python
f = (x**2 - 1) / (x - 1)
f
```

```python
sp.limit(f, x, 1)
```

```python
def f(x):
    y = (x**2 - 1) / (x - 1)
    return y
```

```python
f(0.9999)
```

```python
f(1.5)
```
