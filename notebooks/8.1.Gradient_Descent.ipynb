## 1. Gradient Descent Update Rule & Mathematical Formulation

### Intuition First
Gradient Descent is an iterative first-order optimization algorithm used to find local minima of a differentiable objective function $f(x)$. In machine learning, $f(x)$ represents the Loss Function (e.g. Mean Squared Error or Cross-Entropy). By taking steps proportional to the negative of the gradient, parameter estimates iteratively step downhill toward minimum loss.

### Formal Definition / Math
The parameter update rule at iteration $k+1$ is defined as:

$$\begin{equation}
x^{(k+1)} = x^{(k)} - \eta \cdot \nabla f\left(x^{(k)}\right)
\end{equation}$$

where:
- $x^{(k)}$: Current parameter value vector at iteration $k$.
- $\eta > 0$: Learning rate (step size hyperparameter).
- \nabla f(x^{(k)}): Gradient vector of partial derivatives $\left[ \frac{\partial f}{\partial x_1}, \dots, \frac{\partial f}{\partial x_n} \right]^T$.

#### Worked Manual Derivation Example:
For $f(x) = x^2 \implies f'(x) = 2x$. With initial parameter $x^{(0)} = 10$ and learning rate $\eta = 0.2$:
- **Iteration 1**: $x^{(1)} = 10 - 0.2 \cdot (2 \times 10) = 10 - 4.0 = 6.0$
- **Iteration 2**: $x^{(2)} = 6.0 - 0.2 \cdot (2 \times 6.0) = 6.0 - 2.4 = 3.6$
- **Iteration 3**: $x^{(3)} = 3.6 - 0.2 \cdot (2 \times 3.6) = 3.6 - 1.44 = 2.16$
- **Iteration 4**: $x^{(4)} = 2.16 - 0.2 \cdot (2 \times 2.16) = 2.16 - 0.864 = 1.296$

### Connection to the Code
The code below implements the step-by-step arithmetic sequence for scalar quadratic minimization.

**Key Takeaway**: **Gradient descent iteratively updates parameters in the steepest descent direction $- \eta \nabla f(x)$ to minimize objective loss.**

### Step-by-Step Numerical Walkthrough
- Objective: $y = f(x) = x^2$
- Derivative: $\frac{df}{dx} = 2x$
- Initial value: $x_0 = 10$
- Learning rate: $\eta = 0.2$

- **Iteration 1**: $x_1 = x_0 - \eta \cdot f'(x_0) = 10 - 0.2 \times 20 = 6.0$
- **Iteration 2**: $x_2 = x_1 - \eta \cdot f'(x_1) = 6 - 0.2 \times 12 = 3.6$
- **Iteration 3**: $x_3 = x_2 - \eta \cdot f'(x_2) = 3.6 - 0.2 \times 7.2 = 2.16$
- **Iteration 4**: $x_4 = x_3 - \eta \cdot f'(x_3) = 2.16 - 0.2 \times 4.32 = 1.296$

## 2. Implementing a Generic Iterative Solver

### Intuition First
We modularize Gradient Descent as a flexible Python function that accepts any gradient function, starting point, learning rate, and iteration threshold.

### Connection to the Code
- `def gradient_descent(gradient, start, rate, n_iter)`: Generic algorithm implementation.
- `if np.all(np.abs(a) <= 1e-6): break`: Implements early stopping when parameter updates fall below convergence threshold.
- `g = lambda x: 2 * x`: Gradient function for $f(x) = x^2$.

**Key Takeaway**: **Iterative gradient descent stops automatically when step size updates fall below convergence tolerances.**

```python
import numpy as np
```

```python
def gradient_descent(gradient, start, rate, n_iter):
    s = start
    for i in range(n_iter):
        a = -rate * gradient(s)
        if np.all(np.abs(a) <= 1e-06):
            break
        s += a
    return s
```

```python
g = lambda x: 2 * x
```

```python
# Running gradient descent with moderate learning rate eta = 0.2
r = gradient_descent(gradient=g, start=10.0, rate=0.2, n_iter=7)
r
```

## 3. Learning Rate Hyperparameter Dynamics

### Intuition First
The learning rate $\eta$ is a crucial hyperparameter:
- If $\eta$ is **too large**: The algorithm overshoots the minimum and oscillates or diverges.
- If $\eta$ is **too small**: Convergence is exceedingly slow, requiring thousands of iterations.

### Worked Comparisons:
- **Large Learning Rate ($\eta = 0.8$)**: Overshoots across the origin ($x_0 = 10 \to x_1 = -6 \to x_2 = 3.6$). It oscillates but eventually converges due to quadratic symmetry.
- **Small Learning Rate ($\eta = 0.005$)**: Crawls slowly ($x_0 = 10 \to x_1 = 9.9 \to x_2 = 9.801$), requiring 100 iterations just to reach $3.66$.

**Key Takeaway**: **Selecting an appropriate learning rate is essential to prevent overshooting while maintaining fast convergence.**

```python
# Running gradient descent with large learning rate eta = 0.8 (oscillating step)
r_large = gradient_descent(gradient=g, start=10.0, rate=0.8, n_iter=20)
print(np.round(r_large))
```

```python
# Running gradient descent with small learning rate eta = 0.005 (slow crawling step)
r_small = gradient_descent(gradient=g, start=10.0, rate=0.005, n_iter=100)
r_small
```

## 4. Multivariate Gradient Descent Optimization

### Intuition First
Machine learning models optimize loss across multi-dimensional weight vectors $w \in \mathbb{R}^d$. Multivariate Gradient Descent updates parameter vectors simultaneously along all feature axes using partial derivatives.

### Formal Definition / Math
For $f(x, y) = x^2 + y^4$, the gradient vector is:

$$\begin{equation}
\nabla f(x, y) = \begin{bmatrix} \frac{\partial f}{\partial x} \\ \frac{\partial f}{\partial y} \end{bmatrix} = \begin{bmatrix} 2x \\ 4y^3 \end{bmatrix}
\end{equation}$$

### Connection to the Code
- `g = lambda x: np.array([2 * x[0], 4 * x[1] ** 3])`: Defines multivariate gradient function vector.
- `gradient_descent(g, start=np.array([1.0, 1.0]), rate=0.2, n_iter=50)`: Optimizes parameters simultaneously from starting vector $[1.0, 1.0]^T$.

**Key Takeaway**: **Multivariate gradient descent scales seamlessly to high-dimensional parameter spaces by updating feature weights along gradient vectors.**

```python
# Multivariate gradient descent optimization for f(x, y) = x^2 + y^4
g_multi = lambda x: np.array([2 * x[0], 4 * x[1] ** 3])
sol_multi = gradient_descent(gradient=g_multi, start=np.array([1.0, 1.0]), rate=0.2, n_iter=50)
print(sol_multi)
```
