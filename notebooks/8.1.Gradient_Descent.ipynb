`y = x ** 2`

`dy/dx = 2 * x`

`x0 = 10`

`r = 0.2`

`iteration1 : x1 = x0 - r * d = 10 - 0.2 * (2*10) = 6`

`iteration2 : x2 = x1 - r * d = 6 - 0.2 * (2*6) = 3.59`

`iteration3 : x3 = x2 - r * d = 3.59 - 0.2 * (2*3.59) = 2.154`

`iteration4 : x4 = x3 - r * d = 2.154 - 0.2 * (2*2.154) = 1.29 ...`

```python
import numpy as np
```

```python
def gradient_descent(gradient, start, rate, n_iter):
    s = start
    for i in range(n_iter):
        a = -rate * gradient(s)
        if np.all(np.abs(a) <= 1e-06):
            break
        s += a
    return s
```

```python
g = lambda x: 2 * x
```

```python
r = gradient_descent(gradient=g, start=10.0, rate=0.2, n_iter=7)
r
```
