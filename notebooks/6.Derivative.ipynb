## 1. Concept of Derivatives & Partial Differentiation

### Intuition First
The derivative measures the instantaneous rate of change of a function with respect to an input variable. In machine learning, derivatives drive optimization algorithms (such as Gradient Descent) by indicating the exact direction and steepness needed to adjust model weights to minimize loss.

### Formal Definition / Math
The ordinary derivative of a single-variable function $f(x)$ is defined as the limit of the difference quotient:

$$\begin{equation}
f'(x) = \frac{df}{dx} = \lim_{h \to 0} \frac{f(x + h) - f(x)}{h}
\end{equation}$$

For a multi-variable function $f(x, y)$, the **partial derivative** $\frac{\partial f}{\partial x}$ measures how $f$ changes as $x$ varies while holding all other variables ($y$) constant:

$$\begin{equation}
\frac{\partial f}{\partial x} = \lim_{h \to 0} \frac{f(x + h, y) - f(x, y)}{h}
\end{equation}$$

#### Derivation Steps:
For $f(x, y) = x^3 + 2x + 2y + 3$:
1. Differentiating with respect to $x$ (treating $y$ as a constant):
   $$\frac{\partial f}{\partial x} = \frac{d}{dx}(x^3) + \frac{d}{dx}(2x) + \frac{d}{dx}(2y) + \frac{d}{dx}(3) = 3x^2 + 2 + 0 + 0 = 3x^2 + 2$$
2. Differentiating with respect to $y$ (treating $x$ as a constant):
   $$\frac{\partial f}{\partial y} = \frac{d}{dy}(x^3) + \frac{d}{dy}(2x) + \frac{d}{dy}(2y) + \frac{d}{dy}(3) = 0 + 0 + 2 + 0 = 2$$

### Connection to the Code
- `x = sp.Symbol('x')`, `y = sp.Symbol('y')`: Defines symbolic variables $x$ and $y$.
- `f = x**3 + 2 * x + 3 + 2 * y`: Defines bivariate objective function.
- `sp.diff(f, x)`: Computes partial derivative $\frac{\partial f}{\partial x} = 3x^2 + 2$.
- `sp.diff(f, y)`: Computes partial derivative $\frac{\partial f}{\partial y} = 2$.

### Worked Micro-Example
At point $(x, y) = (2, 5)$:
- Rate of change along $x$-axis: $\frac{\partial f}{\partial x}\Big|_{(2,5)} = 3(2^2) + 2 = 14$.
- Rate of change along $y$-axis: $\frac{\partial f}{\partial y}\Big|_{(2,5)} = 2$.

**Key Takeaway**: **Partial derivatives measure instantaneous sensitivity along individual parameter axes, forming the gradient vector $\nabla f$.**

```python
import sympy as sp
```

```python
x = sp.Symbol("x")
y = sp.Symbol("y")
```

```python
f = x**3 + 2 * x + 3 + 2 * y
f
```

```python
# Partial derivative with respect to x
sp.diff(f, x)
```

```python
# Partial derivative with respect to y
sp.diff(f, y)
```

## 2. The Chain Rule & Trigonometric Functions

### Intuition First
Neural networks are composite functions $f(g(h(x)))$ consisting of multiple nested layers. The **Chain Rule** allows us to calculate how tiny changes at the input propagate through hidden layers to the output—enabling backpropagation.

### Formal Definition / Math
For a composite function $y = f(u)$ where $u = g(x)$:

$$\begin{equation}
\frac{dy}{dx} = \frac{dy}{du} \cdot \frac{du}{dx} = f'(g(x)) \cdot g'(x)
\end{equation}$$

#### Derivation Steps:
For $f(x) = \sin(x^2 + 5x)$:
1. Outer function: $y = \sin(u)$ with $\frac{dy}{du} = \cos(u)$.
2. Inner function: $u = x^2 + 5x$ with $\frac{du}{dx} = 2x + 5$.
3. Applying Chain Rule:
   $$\frac{df}{dx} = \cos(x^2 + 5x) \cdot (2x + 5) = (2x + 5) \cos(x^2 + 5x)$$

### Connection to the Code
- `f = sp.sin(x**2 + 5 * x)`: Formulates composite trigonometric function.
- `sp.diff(f)`: Automatically applies the Chain Rule to produce $(2x + 5)\cos(x^2 + 5x)$.

**Key Takeaway**: **The Chain Rule multiplies outer and inner derivatives, forming the algorithmic engine for Neural Network Backpropagation.**

```python
f = sp.sin(x**2 + 5 * x)
f
```

```python
# Derivative of composite trigonometric function via Chain Rule
sp.diff(f)
```

## 3. Logarithmic Differentiation & Loss Functions

### Intuition First
Logarithmic functions are fundamental in machine learning because Cross-Entropy loss functions ($L = -\sum y \ln(\hat{y})$) and Likelihood estimations rely on logarithms. Differentiating logarithmic terms allows models to calculate gradients for probability distributions.

### Formal Definition / Math
The basic rule for natural logarithm is $\frac{d}{dx}\ln(x) = \frac{1}{x}$. By the Chain Rule, for $f(x) = \ln(g(x))$:

$$\begin{equation}
\frac{d}{dx} \ln(g(x)) = \frac{g'(x)}{g(x)}
\end{equation}$$

#### Derivation Examples:
1. For $f(x) = \ln(x^2)$:
   $$\frac{df}{dx} = \frac{1}{x^2} \cdot (2x) = \frac{2x}{x^2} = \frac{2}{x}$$
2. For $f(x) = \ln(x^2 + 3x - 6)$:
   $$\frac{df}{dx} = \frac{1}{x^2 + 3x - 6} \cdot (2x + 3) = \frac{2x + 3}{x^2 + 3x - 6}$$

### Connection to the Code
- `f = sp.log(x**2)`: Derivative yields `2/x`.
- `f = sp.log(x**2 + 3 * x - 6)`: Derivative yields `(2*x + 3)/(x**2 + 3*x - 6)`.

**Key Takeaway**: **The derivative of $\ln(g(x))$ simplifies to $\frac{g'(x)}{g(x)}$, creating key updates for Cross-Entropy loss optimization.**

```python
f = sp.log(x**2)
f
```

```python
sp.diff(f)
```

```python
f = sp.log(x**2 + 3 * x - 6)
f
```

```python
sp.diff(f)
```

## 4. Exponential Differentiation & Gaussian Gradients

### Intuition First
Exponential functions $e^{g(x)}$ appear throughout probability modeling (Gaussian distributions, Softmax normalization, and RBF kernels). Calculating their derivatives is essential for computing gradient vectors in machine learning algorithms.

### Formal Definition / Math
The exponential function is its own derivative: $\frac{d}{dx} e^x = e^x$. Combined with the Chain Rule, for $f(x) = e^{g(x)}$:

$$\begin{equation}
\frac{d}{dx} e^{g(x)} = g'(x) e^{g(x)}
\end{equation}$$

#### Derivation Step:
For $f(x) = e^{3x^2 - 4x + 9}$:
1. Inner function exponent: $g(x) = 3x^2 - 4x + 9$.
2. Derivative of inner function: $g'(x) = 6x - 4$.
3. Applying Chain Rule:
   $$\frac{df}{dx} = (6x - 4) e^{3x^2 - 4x + 9}$$

### Connection to the Code
- `f = sp.exp(3 * x**2 - 4 * x + 9)`: Defines composite exponential function.
- `sp.diff(f, x)`: Computes derivative `(6*x - 4)*exp(3*x**2 - 4*x + 9)`.

**Key Takeaway**: **The derivative of $e^{g(x)}$ scales the original exponential term by the inner derivative $g'(x)$.**

```python
f = sp.exp(3 * x**2 - 4 * x + 9)
f
```

```python
# Derivative of exponential function via Chain Rule
sp.diff(f, x)
```
