### Solving Linear Systems with SymPy
Here we solve a basic 2D linear system $x - y = -1$ and $3x + y = 9$. We define these equations symbolically and solve them algebraically. This represents finding the intersection of two lines in $\mathbb{R}^2$.

```python
import sympy as sy
```

```python
x, y = sy.symbols('x,y')
```

```python
3 * x + y + x ** 2
```

```python
# x - y = -1 
# 3x + y = 9
```

```python
eq1 = sy.Eq((x - y), - 1)
eq2 = sy.Eq((3 * x + y), 9)
sy.solve((eq1, eq2), (x, y))
```

```python
import numpy as np
```

```python
A = np.array([[1, -1],
              [3, 1]])
```

### Matrix Formulation $Ax = b$
Alternatively, we can express the linear system as a matrix equation $Ax = b$. Here, $A$ is the coefficient matrix and $b$ is the target vector. We can solve for $x$ by calculating $x = A^{-1}b$ or by using numerical solvers, which are generally faster and more stable.

```python
b = np.array([-1, 9])
```

```python
np.linalg.inv(A).dot(b)
```

```python
np.linalg.solve(A, b)
```

```python
from scipy.linalg import solve
solve(A, b)
```

### Visualizing the Linear System
To build geometric intuition, we can plot the two linear equations. The intersection point of the two lines corresponds to the unique solution of the system $Ax = b$.

```python
import matplotlib.pyplot as plt
```

```python
x = np.linspace(-5,5,10)
y1 = x + 1
y2 = 9 - 3*x
plt.plot(x,y1,c='b')
plt.plot(x,y2,c='r')
plt.xlabel('x')
plt.ylabel('y')
plt.axvline(x=2,color='gray',linestyle='--')
plt.axhline(y=3,color='gray',linestyle='--')
```

```python
# x + 3 * y - 2 * z = 5
# 3 * x + 5 * y + 6 * z = 7
# 2 * x + 4 * y + 3 * z = 8
```

```python
x , y , z = sy.symbols('x,y,z')

eq1 = sy.Eq((x + 3 * y - 2 * z), 5)
eq2 = sy.Eq((3 * x + 5 * y + 6 * z), 7)
eq3 = sy.Eq((2 * x + 4 * y + 3 * z), 8)

sy.solve((eq1, eq2, eq3), (x, y, z))
```

### Solving a 3x3 Linear System
We now extend our approach to $\mathbb{R}^3$, solving a system of 3 equations with 3 unknowns. The underlying math remains the same: we formulate $Ax = b$ where $A \in \mathbb{R}^{3 \times 3}$ and use standard numerical solvers to find the intersection of these three planes.

```python
A = np.array([[1, 3, -2],
              [3, 5, 6,],
              [2, 4, 3]])
b = np.array([5, 7, 8])
x = np.linalg.solve(A, b)
x
```
